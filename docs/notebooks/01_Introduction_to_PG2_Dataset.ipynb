# Introduction to the ProteinGym dataset

Welcome to the updated ProteinGym package. This notebook provides an introduction to the dataset package that provides a unified way to access protein data within Protein Gym. Let ProteinGym do the lifting so that you can focus on building your best machine learning model for protein engineering.

ProteinGym solves the issue of dealing with the complex variety of file formats for different protein data streams and allows for sharing, re-using, archiving and versioning of protein engineering datasets.

## What is the ProteinGym dataset?

ProteinGym is an extensive set of Deep Mutational Scanning (DMS) assays and annotated human clinical variants curated to enable thorough comparisons of various mutation effect predictors in different regimes. Both the DMS assays and clinical variants are divided into 1) a substitution benchmark which currently consists of the experimental characterisation of ~2.7M missense variants across 217 DMS assays and 2,525 clinical proteins, and 2) an indel benchmark that includes ∼300k mutants across 74 DMS assays and 1,555 clinical proteins.

The ProteinGym dataset package provides a standardized method for managing these datasets in your machine learning workflows. Importantly it provides:

- **Unified Data Access**: Load protein sequences, structures, assays and MSAs through a single interface.
- **Standardized Format**: Consistent data representation across different protein datasets.
- **Data Archiving**: We maintain versioning of the DMS and ClinVar dataset to share datasets and improve reproducibility
- **Facilitates Benchmarking**: Using the ProteinGym dataset we facilitate the automatic benchmarking of machine learning models in our [ProteinGym-Benchmarking](https://github.com/ProteinGym2/pg2-benchmark)
- **Minimal dependencies**: To keep the package lightweight and prevent package conflicts. We rely on Biotite for structure processing and Biopython for MSA processing on the backend. 

## Pythonic APIs

We provide a pythonic api to allow access to the data as simple as possible

```python
from proteingym.base import Dataset, Manifest

# Load from manifest
manifest = Manifest.from_path("dataset.toml")
dataset = Dataset.from_manifest(manifest)

# Access data
sequences = dataset.sequences
assays = dataset.assays
structures = dataset.structures
msas = dataset.msas
```

## Core Components

The ProteinGym dataset provides an API for the following core components of protein engineering:

### Sequences

- Support for `wild-type`, `starting_sequence` and `engineered_sequence`.
- Support for DNA, RNA and amino acid alphabets
  

### Assays

- Support for experimental measurements linking phenotypes to sequences
- Support for numerical (regression) and catagorical (classification) targets
- Support for tracking assay variables and metadata

### Structures

- Support for experimental and computational structures
- Offloads processing of structures through Biotite, so you can interact with structure objects as you are used to.
- Support for mmcif and binary cif files. 

### Multiple Sequence Alignments

- Support for evolutionary information from related sequences.
- Integration with biotite.sequence.io.fasta.FastaFile MSA class for processing many different MSA formats.
